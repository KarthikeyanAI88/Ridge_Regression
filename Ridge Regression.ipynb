```python
import pandas as pd
import numpy as np
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=(pd.get_dummies(dataset,drop_first=True,dtype=int))
```

```python
dataset
```

```python
dataset.columns
```

```python
independent = dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
```

```python
independent
```

```python
dependent = dataset[['charges']]
```

```python
dependent
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Ridge
param_grid = {"alpha": ['float', 'array-like'], "fit_intercept":['True', 'False'], "max_iter":['None','int'] , "solver":['auto', 'svd', 'cholesky', 'lsqr', 'sparse_cg']}
grid = GridSearchCV(estimator=Ridge, param_grid=param_grid, cv=5, scoring= "neg_mean_squared_error", n_jobs=-1)
grid.fit(independent,dependent)
```

```python
re=grid.cv_results_
print("The R_score value for the best parameter {}:".format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(re)
```

```python
table
```

```python
age_input=float(input("Age:"))
bmi_input=float(input("BMI:"))
children_input=float(input("Children:"))
sex_male_input=int(input("Sex Male 0 or 1:"))
smoker_yes_input=int(input("Smoker Yes 0 or 1:"))
```

```python
Future_Prediction=grid.predict([[age_input,bmi_input,children_input,sex_male_input,smoker_yes_input]])
print("Future_Prediction={}".format(Future_Prediction))
```
